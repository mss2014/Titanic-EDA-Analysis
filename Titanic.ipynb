```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Set seaborn style
sns.set(style="whitegrid")

# Load Titanic train data
df = pd.read_csv("train.csv")
df = pd.read_csv("gender_submission.csv")
```

```python
# Describe all columns including object types
df.describe(include='all')
```

```python
# Dataset info
df.info()
```

```python
print("Survived:\n", df['Survived'].value_counts(), "\n")
print("Sex:\n", df['Sex'].value_counts(), "\n")
print("Pclass:\n", df['Pclass'].value_counts(), "\n")
print("Embarked:\n", df['Embarked'].value_counts(), "\n")
```

```python
#Survival Rate by Gender
sns.barplot(x='Sex', y='Survived', data=df)
plt.title('Survival Rate by Gender')
plt.show()
```

```python
#observations:-
# Survival strongly favors females.
# Males show a low survival rate.
# Gender is a key survival factor.
```

```python
#Survival Rate by Passenger Class
sns.barplot(x='Pclass', y='Survived', data=df)
plt.title('Survival Rate by Passenger Class')
plt.show()
```

```python
#observations:-
# Survival Rate Decreases with Class.
# First Class (Pclass 1) Had Highest Survival.
# Third Class (Pclass 3) Had Lowest Survival.
# Second Class (Pclass 2) Survival is Intermediate.
```

```python
#Age Distribution by Survival
sns.histplot(data=df, x='Age', hue='Survived', multiple='stack', bins=30)
plt.title('Age Distribution by Survival')
plt.show()
```

```python
#observation
# higher survival rates (orange bars) among younger individuals.
# The highest concentration of non-survivors appears to be within the 20-30 age bracket.
# Survival rates (orange) generally decreases with age.
```

```python
#Fare Distribution by Survival
sns.boxplot(x='Survived', y='Fare', data=df)
plt.title('Fare Distribution by Survival')
plt.show()
```

```python
#observations:-
# Survivors (1) generally paid higher fares than those who did not survive (0).
# The fare distribution for survivors is skewed towards higher fares, while the non-survivor fare distribution is more concentrated at lower fares.
```

```python
#Correlation Heatmap
plt.figure(figsize=(10, 6))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```python
#observation:-
# lower passenger classes had lower survival rates.
# Lower class, lower fare.
# Higher fare, more survival.
```
